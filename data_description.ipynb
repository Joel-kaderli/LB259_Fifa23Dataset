```python
import pandas as pd

# Datei mit richtigem Trennzeichen einlesen
df = pd.read_csv("Datensatz_KaderliJoel-Top5Leagues.csv", encoding="utf-8", delimiter=";")

# Spalten anzeigen
print(df.columns)
```

```python


print("Verteilung der Zielvariable (WIN):")
print(df["WIN"].value_counts())
```

```python

print("\nStatistische Kennzahlen für HT_H_Score und HT_A_Score:")
print(df[['HT_H_Score', 'HT_A_Score']].describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,5))
sns.histplot(df['HT_H_Score'], bins=5, kde=True, color="blue", label="Heimteam")
sns.histplot(df['HT_A_Score'], bins=5, kde=True, color="red", label="Auswärtsteam")
plt.legend()
plt.title("Verteilung der Halbzeit-Tore")
plt.xlabel("Tore zur Halbzeit")
plt.ylabel("Häufigkeit")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[['HT_H_Score_Scaled', 'HT_A_Score_Scaled']] = scaler.fit_transform(df[['HT_H_Score', 'HT_A_Score']])

# Überprüfung nach Skalierung
print("\nNach Skalierung:")
print(df[['HT_H_Score_Scaled', 'HT_A_Score_Scaled']].describe())
```
